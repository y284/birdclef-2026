# Perch curves per second

Perch scores every 5 s window with a 1 s stride, so each second is covered by up to 5 windows. Each window's logits
are first transformed, then a second's value is the weighted mean of the windows covering it, with weights
`0.1, 0.3, 0.6, 0.3, 0.1` by its position in the window (the centre second counts most). Two transforms are shown:

- **softmax** over our 203 labels in each window;
- **normalized logits**: per label, the mean over every window (soundscapes and training clips) maps to 0, and the
  99th percentile over the windows of the clips where the label is primary or secondary maps to 1. A label with no
  such clip uses the 99.9th percentile over all soundscape windows instead.

For 10 random labelled soundscapes: up to 6 labelled classes as solid lines, with bars of the same colour under the
curves where they are labelled (per 5 s segment), and the 3 unlabelled classes with the highest values as dashed
lines. For 10 random training clips of at least 15 s, for segment selection: only the clip's primary and secondary
labels, as softmax probability (left) and normalized logit (right).

```python
import ast
import os
from pathlib import Path

if Path.cwd().name == "notebooks":
    os.chdir("..")

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from birdclef import config as C
from birdclef.perch import norm_stats

WEIGHTS = np.array([0.1, 0.3, 0.6, 0.3, 0.1])  # by position of the second inside its 5 s window
N_FILES, N_LABELLED, N_OTHER, MIN_WINDOWS, SEED = 50, 6, 3, 11, 0
```

```python
classes = pd.read_csv(C.PERCH_DIR / "perch_map.csv").primary_label.astype(str).tolist()
rng = np.random.default_rng(SEED)

labels = pd.read_csv(C.RAW_DIR / "train_soundscapes_labels.csv")
for col in ("start", "end"):
    labels[col] = pd.to_timedelta(labels[col]).dt.total_seconds().astype(int)
labels["primary_label"] = labels.primary_label.astype(str).str.split(";")
labels = labels.explode("primary_label")
cached = {p.stem for p in (C.PERCH_DIR / "train_soundscapes").glob("*.npz")}
files = sorted(f for f in labels.filename.unique() if Path(f).stem in cached)
scapes = []
for f in rng.choice(files, min(N_FILES, len(files)), replace=False):
    logits = np.load((C.PERCH_DIR / "train_soundscapes" / f).with_suffix(".npz"))["logits"].astype(np.float32)
    scapes.append({"title": f, "logits": logits, "truth": labels[labels.filename == f], "secondary": set()})

train = pd.read_csv(C.RAW_DIR / "train.csv").astype({"primary_label": str})
clips = []
for i in rng.permutation(len(train)):
    r = train.iloc[i]
    path = (C.PERCH_DIR / "train_audio" / r.filename).with_suffix(".npz")
    if not path.exists():
        continue
    logits = np.load(path)["logits"].astype(np.float32)
    if len(logits) < MIN_WINDOWS:
        continue
    secondary = ast.literal_eval(r.secondary_labels)
    seconds = len(logits) + len(WEIGHTS) - 1
    truth = pd.DataFrame({"primary_label": [r.primary_label, *secondary], "start": 0, "end": seconds})
    clips.append({"title": r.filename, "logits": logits, "truth": truth, "secondary": set(secondary)})
    if len(clips) == N_FILES:
        break
print(f"{len(classes)} Perch classes; {len(scapes)} of {len(files)} labelled soundscapes, {len(clips)} clips")
```

```python
def per_second(x, weights=WEIGHTS):
    """[windows, classes] with a 1 s stride -> [seconds, classes] weighted mean over the covering windows."""
    n = len(x)
    total = np.zeros((n + len(weights) - 1, x.shape[1]))
    norm = np.zeros(n + len(weights) - 1)
    for j, w in enumerate(weights):  # window k covers seconds k .. k + 4; second k + j sits at position j
        total[j : j + n] += w * x
        norm[j : j + n] += w
    return total / norm[:, None]


def softmax(x, t=1.0):
    z = x / t
    z = np.exp(z - z.max(1, keepdims=True))
    return z / z.sum(1, keepdims=True)


MEAN, TOP, fallback = norm_stats(classes)
print(f"{len(fallback)} labels without a positive clip use the soundscape fallback: {[classes[i] for i in fallback]}")


def normalized(x):
    return (x - MEAN) / np.maximum(TOP - MEAN, 1e-6)  # per label: mean -> 0, positive-clip 99th percentile -> 1
```

```python
def plot_curves(items, transform, ylabel, path):
    fig, axes = plt.subplots(len(items), 1, figsize=(13, 3.4 * len(items)), squeeze=False)
    for ax, item in zip(axes[:, 0], items):
        curve = per_second(transform(item["logits"]))
        t = np.arange(len(curve)) + 0.5
        truth = item["truth"]
        labelled = list(dict.fromkeys(truth.primary_label))
        missing = [c for c in labelled if c not in classes]
        shown = sorted((c for c in labelled if c in classes), key=lambda c: -curve[:, classes.index(c)].max())
        hidden, shown = shown[N_LABELLED:], shown[:N_LABELLED]
        others = [classes[i] for i in np.argsort(-curve.max(0)) if classes[i] not in labelled][:N_OTHER]
        drawn = curve[:, [classes.index(c) for c in shown + others]]
        lo, hi = min(drawn.min(), 0), max(drawn.max(), 1)
        step = 0.06 * (hi - lo)  # vertical spacing of the label bars, below every curve
        for i, c in enumerate(shown + others):
            color = plt.cm.tab10(i % 10)
            y = curve[:, classes.index(c)]
            if c in shown:
                ax.plot(t, y, color=color, lw=1.6, label=f"{c} (secondary)" if c in item["secondary"] else c)
                bar = truth[truth.primary_label == c]  # one row of bars per class
                ax.hlines([lo - step * (i + 1)] * len(bar), bar.start, bar.end, color=color, lw=3)
            else:
                ax.plot(t, y, color=color, lw=1.1, ls="--", label=f"{c} (not labelled)")
        ax.axhline(0, color="grey", lw=0.6)
        ax.set_xlim(0, len(curve))
        ax.set_ylim(lo - step * (len(shown) + 1), hi + 0.02 * (hi - lo))
        ax.set_yticks([v for v in ax.get_yticks() if lo <= v <= hi])
        ax.set_ylabel(ylabel)
        notes = [f"+{len(hidden)} labelled not drawn"] * bool(hidden)
        notes += [f"labelled, not in Perch: {', '.join(missing)}"] * bool(missing)
        ax.set_title("   ".join([item["title"], *notes]), fontsize=9, loc="left")
        ax.legend(fontsize=7, loc="upper left", bbox_to_anchor=(1.0, 1.0), frameon=False)
    axes[-1, 0].set_xlabel("second")
    fig.tight_layout()
    Path(path).parent.mkdir(exist_ok=True)
    fig.savefig(path, dpi=110, bbox_inches="tight")
    plt.show()
```

## Training clips

```python
def plot_clips(items, path):
    panels = [(softmax, "probability"), (normalized, "normalized logit")]
    fig, axes = plt.subplots(len(items), 2, figsize=(15, 2.8 * len(items)), squeeze=False)
    for row, item in zip(axes, items):
        labelled = [c for c in item["truth"].primary_label if c in classes]
        for ax, (transform, ylabel) in zip(row, panels):
            curve = per_second(transform(item["logits"]))
            t = np.arange(len(curve)) + 0.5
            for i, c in enumerate(labelled):
                kind = "secondary" if c in item["secondary"] else "primary"
                ax.plot(t, curve[:, classes.index(c)], color=plt.cm.tab10(i % 10), lw=1.6, label=f"{c} ({kind})")
            ax.axhline(0, color="grey", lw=0.6)
            ax.set_xlim(0, len(curve))
            ax.set_ylabel(ylabel)
            if transform is softmax:
                ax.set_ylim(0, 1.02)
        missing = [c for c in item["truth"].primary_label if c not in classes]
        note = f"   not in Perch: {', '.join(missing)}" if missing else ""
        row[0].set_title(item["title"] + note, fontsize=9, loc="left")
        row[1].legend(fontsize=7, loc="upper left", bbox_to_anchor=(1.0, 1.0), frameon=False)
    for ax in axes[-1]:
        ax.set_xlabel("second")
    fig.tight_layout()
    fig.savefig(path, dpi=110, bbox_inches="tight")
    plt.show()


plot_clips(clips, "results/perch_curves_clips.png")
```
